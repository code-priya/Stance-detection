```python
import pandas as pd
from sklearn.model_selection import train_test_split

# ==========================
# Load datasets
# ==========================
merged_df = pd.read_csv("merged_hindi.csv")
labelled_df = pd.read_csv("Hindi_500_labeled.csv")

# ==========================
# Check required columns
# ==========================
assert list(merged_df.columns) == list(labelled_df.columns), \
    "Both datasets must have the same columns."

# ==========================
# Random seed for reproducibility
# ==========================
SEED = 42

# ==========================
# Pick 20% (100 samples) from labelled dataset
# ==========================
labelled_subset = labelled_df.sample(
    frac=0.20,
    random_state=SEED
).reset_index(drop=True)

# ==========================
# Merge datasets
# ==========================
final_df = pd.concat(
    [merged_df, labelled_subset],
    ignore_index=True
)

# Remove duplicates if any
final_df = final_df.drop_duplicates(subset="text").reset_index(drop=True)

print(f"Final dataset size: {len(final_df)}")

# ==========================
# 80/10/10 Stratified Split
# ==========================
train_df, temp_df = train_test_split(
    final_df,
    test_size=0.20,
    stratify=final_df["label"],
    random_state=SEED
)

valid_df, test_df = train_test_split(
    temp_df,
    test_size=0.50,
    stratify=temp_df["label"],
    random_state=SEED
)

# ==========================
# Save splits
# ==========================
train_df.to_csv("train_hindi.csv", index=False)
valid_df.to_csv("valid_hindi.csv", index=False)
test_df.to_csv("test_hindi.csv", index=False)

# ==========================
# Display statistics
# ==========================
print("\nDataset Split")
print("-" * 40)
print(f"Train      : {len(train_df)}")
print(f"Validation : {len(valid_df)}")
print(f"Test       : {len(test_df)}")
print(f"Total      : {len(final_df)}")

print("\nTrain Label Distribution")
print(train_df["label"].value_counts(normalize=True))

print("\nValidation Label Distribution")
print(valid_df["label"].value_counts(normalize=True))

print("\nTest Label Distribution")
print(test_df["label"].value_counts(normalize=True))
```
